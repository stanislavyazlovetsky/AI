7. Визначити функцію, яка приймає на вході масив і транспонує його. Скористайтесь методом reshape. І не забудьте почати з docstrings.

```python
def transpose_with_reshape(array):
    if len(array.shape) != 2:
        raise ValueError("Масив має бути двовимірним.")
    rows, cols = array.shape
    return array.reshape(cols, rows).T
array3 = np.array([[1, 2, 3], [4, 5, 6]])
transposed_array = transpose_with_reshape(array3)

print("Оригінальний масив:")
print(array3)

print("\nТранспонований масив:")
print(transposed_array)
```

```python
array4 = np.array([[1, 2, 3], [4, 5, 6]])
array5 = np.array([[6, 5, 4], [3, 2, 1]])
elementwise_addition = array4 + array5
elementwise_subtraction = array4 - array5
scalar_multiplication = array4 * 2
elementwise_multiplication = array4 * array5
matrix_multiplication = np.dot(array4, array5.T)
print("Масив 4:")
print(array4)
print("\nМасив 5:")
print(array5)
print("\n1) По-елементне додавання:")
print(elementwise_addition)
print("\n2) По-елементне віднімання:")
print(elementwise_subtraction)
print("\n3) Множення масиву на число (2):")
print(scalar_multiplication)
print("\n4) По-елементне множення:")
print(elementwise_multiplication)
print("\n5) Матричне множення:")
print(matrix_multiplication)
```

20. Видалити колонки, де кількість непустих значень становить менше 30% від загальної кількості записів.

21. Очистити датафрейм від дублікатів (дублюючихся рядків) та записів, де атака продовжувалась відʼємну або нульову кількість часу (що фізично неможливо). Не забудьте оновити індекс. Скільки рядків було видалено? Як ви це перевірили у коді?

```python
df["time_start"] = pd.to_datetime(df["time_start"], errors='coerce')
df["time_end"] = pd.to_datetime(df["time_end"], errors='coerce')
print("Кількість некоректних дат у time_start:", df["time_start"].isna().sum())
print("Кількість некоректних дат у time_end:", df["time_end"].isna().sum())
df = df.dropna(subset=["time_start", "time_end"])
df["duration"] = (df["time_end"] - df["time_start"]).dt.total_seconds() / 3600
rows_before = len(df)
df = df[df["duration"] > 0]
df = df.reset_index(drop=True)
rows_after = len(df)
print("Кількість видалених рядків:", rows_before - rows_after)
```
