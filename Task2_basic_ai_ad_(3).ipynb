````python
import pandas as pd
df = pd.read_csv("titanic.csv")
import matplotlib.pyplot as plt
import seaborn as sns
````

**Задача 1:** Який відсоток виживання серед чоловіків / жінок? 

````python
fig, axes = plt.subplots(1, 2, figsize=(12, 5))
survived_counts = df["Survived"].value_counts()
axes[0].pie(
    survived_counts,
    labels=survived_counts.index,
    autopct="%1.1f%%",
    colors=["#4C72B0", "#DD8452"],
    startangle=140,
    wedgeprops={"edgecolor": "black"},
    explode=[0.05, 0],
)
axes[0].set_title("Survived")

sns.countplot(x="Survived", data=df, palette=["#4C72B0", "#DD8452"], ax=axes[1])
axes[1].set_title("Survived")
axes[1].set_xlabel("Survived")
axes[1].set_ylabel("Count")

plt.tight_layout()
plt.show()
````

**Задача 4:** Назвіть найпопулярніше **жіноче** ім'я на кораблі. 

````python
df_female = df[df["Sex"] == "female"]

df_female["First Name"] = df_female["Name"].str.extract(r"(\b[A-Za-z]+)\b", expand=False)

most_common_female_name = df_female["First Name"].mode()[0]
print("Найпопулярніше жіноче ім'я:", most_common_female_name)
````

**Задача 5:** Назвіть найпопулярніше **чоловіче** ім'я на кораблі. 

````python
df_male = df[df["Sex"] == "male"]

df_male["First Name"] = df_male["Name"].str.extract(r"(\b[A-Za-z]+)\b", expand=False)

most_common_male_name = df_male["First Name"].mode()[0]
print("Найпопулярніше чоловіче ім'я:", most_common_male_name)
````

**Задача 8** Побудуйте наступні діаграми розподілів:

image.png

````python
df = df.dropna(subset=['Age'])

plt.figure(figsize=(12, 10))

plt.subplot(2, 2, 1)
df[df['Survived'] == 0]['Age'].hist(bins=30, alpha=0.7, color='teal')
plt.title("Survived = 0")
plt.xlabel("Age")
plt.ylabel("Frequency")

plt.subplot(2, 2, 2)
df[df['Survived'] == 1]['Age'].hist(bins=30, alpha=0.7, color='salmon')
plt.title("Survived = 1")
plt.xlabel("Age")
plt.ylabel("Frequency")

plt.subplot(2, 1, 2)
sns.histplot(df[df['Survived'] == 0]['Age'], bins=30, kde=False, color='blue', label='Died', alpha=0.6)
sns.histplot(df[df['Survived'] == 1]['Age'], bins=30, kde=False, color='red', label='Survived', alpha=0.6)
plt.legend()
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.title("Age Distribution of Deceased and Survived Passengers")

plt.tight_layout()
plt.show()
````

***Завдача 10** Побудуйте діаграму розподілу ```вартості квитка``` відповідно до якого классу належіть пасажир.

````python
sns.set_style("whitegrid")

plt.figure(figsize=(8, 5))
sns.barplot(x=df["Pclass"], y=df["Fare"], estimator=lambda x: sum(x)/len(x), palette="pastel")

plt.xlabel("Клас пасажира", fontsize=12)
plt.ylabel("Середня вартість квитка", fontsize=12)
plt.title("Середня вартість квитка для кожного класу", fontsize=14)

plt.show()
````
